# **Chain In LangChain**

```python
from langchain_google_genai import ChatGoogleGenerativeAI

import os 
from dotenv import load_dotenv
# This function will load all the variables from .env file into the environment
# make them available to in the os.environ dictionary
load_dotenv()   

if os.environ.get("GOOGLE_API_KEY"):
    print ("GOOGLE_API_KEY found in the environment variables")
else:
    raise ValueError ("GOOGLE_API_KEY not found in the environment variables")


from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

# for the string out praser 
from langchain_core.output_parsers import StrOutputParser


llm_gemini = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=0)
```

# **First Chain**

```python
# Task-1 [prompt] 

prompt_template = ChatPromptTemplate.from_messages([
    ("system","you are a helpful assistant"),
    ("human","{input}")
])
```

```python
# Task-2 [LLM]

llm_gemini = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=0)
```

# in this .content show only the necessary things and don't show unnecessary content such as parameter,token size
# we can use .content or built-in function in langchain which called String output praser 
# what it will do : String output praser :
# it will take your Generated response and it will only fetch the content out of it and behind the scenes 
# it is doing the same things it is writing the . content but it is automating that 

```python
# Task -3 [String Parser]

Str_parser = StrOutputParser() # create a instance of the function of String output praser as Str_parse
```

# **Manual Invocation**

```python
template = ChatPromptTemplate.invoke({"input":"what is the capital of France"})
# above we invoke the input 
res =llm_gemini.invoke(template) # Here we invoke the LLM 
final_result = res.content # here we will see the unnecessary content 
```

# **Chain Invocation**

```python
# there are two way to create chain which mean invoke the chain 
# pipe operator which is also known LCEL => langchain expression language 


# this below code is how the chain working first start with prompt_template  then to llm_gemini as the llm and last to string output praser
chain = prompt_template | llm_gemini | Str_parser

# to pass the input to chain by invoke 
chain.invoke({"input":"what is the capital of France"})

# this running automatically all the process 
```

# anthter way to create chain (two)

```python
# using runnable 
from langchain_core.runnables import RunnableSequence

chain_1 = RunnableSequence(prompt_template, llm_gemini, Str_parser)

chain_1.invoke({"input":"what is the capital of France"})
```
