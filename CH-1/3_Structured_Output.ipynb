# **with LLM in pydantic**

```python
from pydantic import BaseModel , Field
# fields are used to define the type of the field or the type of the value that the field will hold or 
#fields will give the LLM what it want to generate in that field

class joke(BaseModel):
    setup: str = Field(description="The setup for the joke")
    punchline: str = Field(description="The punchline for the joke")    
```

```python
llm_structured_output = llm_gemini.with_structured_output(joke)
llm_structured_output.invoke("What is the best joke about python?")

#joke(setup='Why do Python programmers prefer dark mode?', punchline='Because light attracts bugs!')
#llm_structured_output.invoke("What is the best joke about python?").content

# here content is not a method  because it is a property of the joke object or 
# .content cannont be used for the pydantic object joke 

#The real reason IS:
#Because the returned object is a Pydantic model, not an LLM message.

#content cannot be used because llm_structured_output.invoke() returns a Pydantic object (Joke), not a chat message.
#A Joke object only has fields defined in the schema (setup, punchline), and it does not contain a .content attribute
```

```python
result = llm_structured_output.invoke("What is the best joke about python?")
type(result)
# __main__.joke is the pydantic model object 
result.setup
result.punchline
```

# **with LLM in TypedDict**

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup:str
    punchline:str
```

# TypedDict as classteacher who is less strict and the pydantic act as the principle who is more strict in checking 

```python
obj = llm_schema_td({"setup":"some setup","punchline":"some punchline"})
obj 
# output :{'setup': 'some setup', 'punchline': 'some punchline'}
obj1 = llm_schema_td({"ketchup":"some setup","punchline":"some punchline"})
obj1
# output :{'ketchup': 'some setup', 'punchline': 'some punchline'}

# it will not show error like pydantic 
```

```python
llm_structured_typed_dict = llm_gemini.with_structured_output(llm_schema_td)
result = llm_structured_typed_dict.invoke("tell me joke")
result 

```
