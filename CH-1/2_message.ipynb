```python
from langchain_google_genai import ChatGoogleGenerativeAI

import os 
from dotenv import load_dotenv
# This function will load all the variables from .env file into the environment
# make them available to in the os.environ dictionary
load_dotenv()   

if os.environ.get("GOOGLE_API_KEY"):
    print ("GOOGLE_API_KEY found in the environment variables")
else:
    raise ValueError ("GOOGLE_API_KEY not found in the environment variables")
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

from langchain_google_genai import ChatGoogleGenerativeAI
llm_gemini = ChatGoogleGenerativeAI(model="gemini-2.0-flash", temperature=0)
messages =[
    # below we can mention how the model should behave or tone and talk like a gen-z assistant 
    SystemMessage(content="You are a gen-z assistant who always answers in a fun and engaging way."),
    HumanMessage(content="Hello, tell me a fun fact ")
]
llm_gemini.invoke(messages).content
```

# **Prompt**

Prompt are the messages that we send to the LLM.
The message are more static way and the prompt are dynamic way to send the messages to the LLM.
Prompt are more user-friendly than messages. 

# **without set the tone or behavior  like above system message we will use the PromptTemplate** 

```python
from langchain_core.prompts import PromptTemplate
# this input 
user_input = input("Enter a topic for fun fact : ")
# dynamic prompt instead of static prompt with PromptTemplate.from_template(content{input})
dynamic_prompt = PromptTemplate.from_template("Tell me a fun fact about {topic}")
#without injecting the input or not passing the input 
dynamic_prompt
#PromptTemplate(input_variables=['topic'], input_types={}, partial_variables={}, template='Tell me a fun fact about {topic}')
# with injecting the input or passing the input 
ready_prompt =dynamic_prompt.invoke({"topic": user_input})
#ouptut:#StringPromptValue(text='Tell me a fun fact about flowers')

llm_gemini.invoke(ready_prompt)  # this used .invoke() method to call the llm or run the llm 

llm_gemini.invoke(ready_prompt).content # this will  get the content only
```

# with set the tone or behavior  like above system message we will use the ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate

Prompt_Template = ChatPromptTemplate.from_messages(
    [
        ("system","You are a polite assistant"),
        ("user","write a fun fact about {topic}")
    ]
)
user_input = input("Enter the topic :")
ready_prompt = Prompt_Template.invoke({"topic":user_input})
# ready_prompt.messages

llm_gemini.invoke(ready_prompt.messages).content
 
```

```python
from langchain_core.prompts import ChatPromptTemplate

Prompt_Template = ChatPromptTemplate.from_messages(
    [
        ("system","You are a {tone} assistant"),
        ("user","write a fun fact about {topic}")
    ]
)
user_input = input("Enter the topic :")
user_tone = input("Enter the tone :")
ready_prompt = Prompt_Template.invoke({"topic":user_input ,"tone":user_tone})
# ready_prompt.messages

llm_gemini.invoke(ready_prompt.messages).content

# .messages removes the Python wrapper and gives only Ai -readable text.
 
```
